# 1. Vector/Matrix Definition and Usage

## 1.1 Vector Representation

```python
class Vector:
    def __init__(self, x = 0.0, y=0.0):
        self.x = x
        self.y = y

    def __str__(self):
        return "[{}, {}]".format(str(self.x), str(self.y))
```

```python
a = Vector(2, 4)
```

```python
print(a)
```

```python
b = Vector(6, 3)
print(b)
```

```python
def add(self, vector2):
    c = Vector()

    c.x = self.x + vector2.x
    c.y = self.y + vector2.y

    return c

Vector.add = add
```

```python
c = a.add(b)
print(c)
```

```python
def mul(self, scaler):
    return Vector(self.x * scaler, self.y * scaler)

Vector.mul = mul
```

```python
c_mul = a.mul(2)
print(c_mul)
```

```python
def sub(self, vector2):
    return self.add( vector2.mul(-1) )

Vector.sub = sub
```

```python
d_min_vector2 = c_mul.sub(c)
print(d_min_vector2)
```

## 1.2 Matrix Representation (Dense)

```python
class Matrix:

    def __init__(self, dims, fills):
        self.rows = dims[0]
        self.cols = dims[1]

        self.A = [ [fills] * self.cols for i in range(self.rows) ]

    def __str__(self):

        rows = len(self.A)
        ret = ""

        for i in range(rows):
            cols = len(self.A[i])

            for j in range(cols):
                ret += str(self.A[i][j]) + "\t"

            ret += "\n"

        return ret
```

```python
m = Matrix((3, 4), 2.0)
```

```python
print(m)
```

## 1.3 Matrix Representation (Sparse)

```python
class Matrix:

    def __init__(self, dims):
        self.rows = dims[0]
        self.cols = dims[1]
        self.vals = {}

    def get(self, i, j):

        if i < 0 or i > self.rows:
            raise ValueError("Row index out of range")
        if j < 0 or j > self.cols:
            raise ValueError("Columns index out of range")

        if (i, j) in self.vals:
            return self.vals[(i, j)]

        return 0.0


    def set(self, index, val):
        self.vals[index] = val

    def __str__(self):
        rows = self.rows
        cols = self.cols

        ret = ""
        for i in range(rows):

            for j in range(cols):

                if (i, j) in self.vals:
                    ret += str(self.vals[(i, j)]) + "\t"
                    continue
                
                ret += str(0.0) + "\t"

            ret += "\n"

        return ret
            
```

```python
m_sparse = Matrix((5, 5))
```

```python
print(m_sparse.vals)
```

```python
m_sparse.get(1, 4)
```

```python
m_sparse.set((1, 4), 10)
```
