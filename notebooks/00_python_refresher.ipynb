```python
def good_enough(x, guess):
    return abs((guess * guess)-x) < 0.00001
    
def improve_guess(x, guess):
    return (guess + x/guess)/2
    
def sqrt(x, guess=0.001):
    if good_enough(x, guess):
        return guess
    else:
        return sqrt(x, improve_guess(x, guess))
```

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
    def __str__(self): # this is a dunder function
        return "[" + str(self.x) + " , " + str(self.y) + "]"
```

## 1. Composition

```python
class Shape:
    def __init__(self, points):
        self.points = points
    
    def __str__(self):
        ret = ""

        for i in self.points:
            ret += str(i) + " - "

        return ret
```

```python
p1 = Point(5, 5)
p2 = Point(10, 4)
p3 = Point(5, 10)
p = [p1, p2, p3]

sh = Shape(p)
print(sh)
```

## Adding fucntion in class later

```python
def print_points(self):
    for i in self.points:
        print(i)

Shape.print_points = print_points
```

```python
sh.print_points()
```

## 2. Inheritance

```python
class Triangle(Shape):
    def is_triangle(self):
        if (len(self.points) == 3):
            return True
        else:
            return False
```

```python
t = Triangle(p)
```

```python
print(t)
```

```python
t.print_points()
```

```python
def get_area(self): 
    vertices = self.points 
    n = len(vertices) 
    a = 0.0 
    for i in range(n): 
        j = (i + 1) % n 
        a += (vertices[i].x * vertices[j].y - vertices[j].x * vertices[i].y) 
    return abs(a) / 2.0 

Triangle.get_area = get_area
```

## 3 Access Parent Class Overridden Methods

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * self.length + 2 * self.width

    def __str__(self):
        return "L: "+ str (self.length) + "  W: " + str(self.width)
```

```python
rect = Rectangle(2, 4)
print(rect)
```

```python
class Square(Rectangle):
    def __init__(self, length):
        super().__init__(length, length)

    def __str__(self):
        return "Square: " + super().__str__()
```

```python
square = Square(4)
square.area()
```
