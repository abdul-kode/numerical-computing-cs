# Dot Producct

```python
A = [2, 4, 6]
B = [1, 3, 5]

dot_product = sum(i*j for i, j in zip(A, B))
dot_product
```

# Cross Product

```python
A = [2, 4, 6]
B = [1, 3, 5]

cross_product = [
    (A[1] * B[2]) - (A[2] * B[1]),  # X component
    (A[2] * B[0]) - (A[0] * B[2]),  # Y component
    (A[0] * B[1]) - (A[1] * B[0]),  # Z component
]

cross_product    
```
