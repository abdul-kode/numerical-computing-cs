```python
class Vector:
    def __init__(self, x = 0.0, y=0.0):
        self.x = x
        self.y = y

    def __str__(self):
        return "[{}, {}]".format(str(self.x), str(self.y))

    def add(self, vector2):
        return Vector(self.x + vector2.x, self.y + vector2.y)

    def mul(self, scaler):
        return Vector(self.x * scaler, self.y * scaler)

    def sub(self, vector2):
        return self.add(vector2.mul(-1))
```

```python
a = Vector(2, 4)
b = Vector(4, 2)
c_add = a.add(b)
scaler = 2
c_mul = a.mul(scaler)
c_sub = b.sub(a)

print(f"Vector a: {a}")
print(f"Vector b: {b}\n")
print(f"a + b: {c_add}")
print(f"a * {scaler}: {c_mul}")
print(f"b - a: {c_sub}")
```

```python
class Vector3:
    def __init__(self, x=0.0, y=0.0, z=0.0):
        self.x = x
        self.y = y
        self.z = z

    def __str__(self):
        return f"({self.x}, {self.y}, {self.z})"
        
    def add(self, other):
        return Vector3(self.x + other.x, self.y + other.y, self.z + other.z)

    def mul(self, scalar):
        return Vector3(self.x * scalar, self.y * scalar, self.z * scalar)
    
    def sub(self, other):
        return Vector3(self.add(other.mul(-1)))
```

# 1. Dot Product

### 1.1 Two Elements

```python
def dot(self, vector2):
    
    temp = Vector(self.x * vector2.x, self.y * vector2.y)
    
    return sum(list((temp.x, temp.y)))

Vector.dot = dot
```

### 1.2 Two Elements

```python
def dot(self, vector2):

    temp = Vector3(self.x * vector2.x, self.y * vector2.y, self.z * vector2.z)

    return sum(list((temp.x, temp.y, temp.z)))

Vector3.dot = dot
```

```python
# --- 2D Vector Demonstration ---
v1 = Vector(3, 4)
v2 = Vector(1, 2)

v_add = v1.add(v2)   # (4, 6)
v_sub = v1.sub(v2)   # (2, 2)
v_mul = v1.mul(3)    # (9, 12)
v_dot = v1.dot(v2)   # (3*1) + (4*2) = 11

print(f"Vector 2D Dot Product: {v_dot}")


# --- 3D Vector Demonstration ---
v3 = Vector3(1, 3, 5)
v4 = Vector3(2, 4, 6)

v3_dot = v3.dot(v4)  # (1*2) + (3*4) + (5*6) = 2 + 12 + 30 = 44

print(f"Vector 3D Dot Product: {v3_dot}")
```
