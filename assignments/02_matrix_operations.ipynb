# 1. Matrix Addition of Sparse

```python
class Matrix:

    def __init__(self, dims):
        self.rows = dims[0]
        self.cols = dims[1]
        self.vals = {}

    def get(self, i, j):

        if i < 0 or i > self.rows:
            raise ValueError("Row index out of range")
        if j < 0 or j > self.cols:
            raise ValueError("Columns index out of range")

        if (i, j) in self.vals:
            return self.vals[(i, j)]

        return 0.0


    def set(self, index, val):
        self.vals[index] = float(val)

    def __str__(self):
        rows = self.rows
        cols = self.cols

        ret = ""
        for i in range(rows):

            for j in range(cols):

                if (i, j) in self.vals:
                    ret += str(self.vals[(i, j)]) + "\t"
                    continue
                
                ret += str(0.0) + "\t"

            ret += "\n"

        return ret
            
```

# 2. Sparse Matrix Multiplication

```python
def mul(self, scaler):

    temp_matrix = Matrix((self.rows, self.cols))
    
    for i in self.vals:
        temp_matrix.vals[i] = self.vals[i] * scaler

    return temp_matrix

Matrix.mul = mul
```

```python
print(f"Before:\n{m_sparse3}")
m_sparse4 = m_sparse3.mul(2)
print("After: \n", m_sparse4)
```

# 3. Sparse Matrix Subtraction

```python
def sub(self, matrix2):
    temp_matrix = Matrix((self.rows, self.cols))

    for i in matrix2.vals.keys():
        new_val = matrix2.vals[i] * -1
        temp_matrix.set(i, new_val)

    return self.add(temp_matrix)

Matrix.sub = sub
```
